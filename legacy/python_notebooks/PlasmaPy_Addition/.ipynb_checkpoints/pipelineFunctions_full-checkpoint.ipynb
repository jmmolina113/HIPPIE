```python
import numpy as np
import h5py
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter

import warnings
warnings.filterwarnings("ignore",category = RuntimeWarning)

# Custom function definitions

def align_pinholes_2d_v2(input_pinhole1, input_pinhole2, where_to_look, average_range, maxima_coords):
    import numpy as np
    iteration = 0
    keep_going = True
    net_shift_vert = 0
    net_shift_horz = 0
    vert_shifts = [0]  # Net vertical shift applied to pinhole 2
    horz_shifts = [0]  # Net horizontal shift applied to pinhole 2
    stuck = [0, 0]
    manual_shift = [0, 0]

    while keep_going:
        iteration += 1

        # Call to external function
        pinhole1_aligned, pinhole2_aligned, keep_going, shift_horz, shift_vert, target_edge = iterative_pinhole_alignment_2d_v2(
            input_pinhole1, input_pinhole2, where_to_look, average_range, maxima_coords, stuck, manual_shift
        )
        input_pinhole2 = pinhole2_aligned

        vert_shifts.append(shift_vert)  # Net vertical shift applied to pinhole 2
        horz_shifts.append(shift_horz)  # Net horizontal shift applied to pinhole 2

        stuck[0], manual_shift[0] = is_stuck(vert_shifts)
        stuck[1], manual_shift[1] = is_stuck(horz_shifts)

        if stuck[0] == 2 or stuck[1] == 2:
            raise ValueError("Alignment Failed")

    net_shift_horz = np.sum(horz_shifts)
    net_shift_vert = np.sum(vert_shifts)

    return pinhole1_aligned, pinhole2_aligned, net_shift_vert, net_shift_horz, target_edge

def background_2d(pinhole, maxima_coords):
    import numpy as np
    import matplotlib.pyplot as plt

    slice_ = pinhole[:, maxima_coords[1] - 50:maxima_coords[1] + 50]  # Isolating a slice around the maxima
    
    lineout = np.nanmean(slice_, axis=1)  # Creating a lineout with the non-nan values
    #lineout = np.flip(lineout)  # Flipping the lineout to make the data easier

    indices = np.isnan(lineout)  # Picking out the NaN values
    lineout = lineout[~indices]
    grad_lineout = np.abs(np.gradient(lineout))  # Taking the absolute of the gradient of the lineout

    iteration = 0  # Initializing the while loop
    background = []

    while iteration < len(grad_lineout) and grad_lineout[iteration] < 0.2:
        background.append(lineout[iteration])  # Keeping values in the lineout that satisfy the controlling expression
        iteration += 1  # Incrementing the iteration
    
    if len(background) == 0:
        reduced_image_lineout = np.nanmean(pinhole, axis=1, where=~np.isnan(pinhole))
        lineout_to_use = np.flip(reduced_image_lineout)
        background = np.nanmean(lineout_to_use[:10])
    else:
        background = np.nanmean(background)
        
    
    return background

def binning_algorithm_v2(pinhole, bin_x, bin_y, what_to_do):
    
    from scipy.stats import gmean
    if bin_x == 1 and bin_y == 1:
        return pinhole

    # Initialize the binning algorithm
    rows, cols = pinhole.shape
    col_anchors = np.arange(bin_x, cols + bin_x, bin_x)
    row_anchors = np.arange(bin_y, rows + bin_y, bin_y) - 1

    if row_anchors[-1] != (rows - 1):
        row_anchors = np.append(row_anchors, rows - 1)

    if col_anchors[-1] != cols:
        col_anchors = np.append(col_anchors, col_anchors[-1] + bin_x)

    # Create the reduced image
    reduced_pinhole = np.empty((len(row_anchors), len(col_anchors)))
    reduced_pinhole[:] = np.nan

    for new_row in range(len(row_anchors)):
        for new_col in range(len(col_anchors)):
            bin_values = np.full((bin_y, bin_x), np.nan)
            cols_to_use = np.arange(col_anchors[new_col] - (bin_x - 1), col_anchors[new_col] + 1)
            rows_to_use = np.arange(row_anchors[new_row] - (bin_y - 2), row_anchors[new_row] + 1)

            for i in range(bin_x):
                for j in range(bin_y):
                    if (cols_to_use[i] >= cols) or (rows_to_use[j] >= rows):
                        bin_values[j, i] = np.nan
                    else:
                        bin_values[j, i] = pinhole[rows_to_use[j], cols_to_use[i]]

            if what_to_do == "mean":
                reduced_pinhole[new_row, new_col] = np.nanmean(bin_values)
            elif what_to_do == "median":
                reduced_pinhole[new_row, new_col] = np.nanmedian(bin_values)
            elif what_to_do == "geomean":
                reduced_pinhole[new_row, new_col] = gmean(bin_values[~np.isnan(bin_values)])

    return reduced_pinhole



def calculate_enclosed_area(lineout1, lineout2):
    import numpy as np
    from scipy.integrate import trapz
    
    difference = np.abs(lineout1 - lineout2)

    upper_index = 500
    while not np.isnan(difference[upper_index]):
        upper_index += 1
        if upper_index == 1051:
            upper_index = len(difference) - 1
            break

    lower_index = 500
    while not np.isnan(difference[lower_index]):
        lower_index -= 1
        if lower_index == 0:
            lower_index = 0
            break

    area = np.abs(trapz(difference[lower_index + 1:upper_index]))
    return area


def calculate_temperature(R, S, temp_bound):
    import numpy as np
    import matplotlib.pyplot as plt
 
    
    T = np.logspace(2, 4, 1001)  # Defining temperature range
    Te = np.full(S.shape, np.nan)

    for i in range(S.shape[0]):
        for j in range(S.shape[1]):
            Te[i, j] = find_temp(R, S[i, j], T)  # Finding
            
    return Te

def calculate_theoretical_ratio(energies, temps, trans_coeffs):
    import numpy as np
    from scipy.special import kv  # Modified Bessel function of the second kind
    from scipy.integrate import trapz
    
    # Initialize Integral array

    integral = np.zeros(len(temps))
    W = trans_coeffs  # Filter function

    for i, temp in enumerate(temps):
        B = kv(0, energies / (2 * temp))  # Modified Bessel function of the second kind (order 0)
        expo = np.exp(-energies / (2 * temp))  # Exponential factor
        K = detector_response(energies)  # Detector Response function (needs to be defined elsewhere)
        
        integrand = B * expo * W * K  # Product to integrate over
        integral[i] = trapz(integrand, energies)  # Numerical integration using trapezoidal rule
    
    return integral

def calculate_transmission_curves(filter_file_name1, filter_file_name2, filter_thickness1, filter_thickness2):
    import numpy as np

    if isinstance(filter_file_name2, (int, float)):
        # Single filter
        energy_data, transmission_curve_full = get_filter_transmission_v2(filter_file_name1, filter_thickness1, 0.1)
    else:
        # Multiple filters
        energy_data, transmission_curve1_full = get_filter_transmission_v2(filter_file_name1, filter_thickness1, 0.1)
        _, transmission_curve2_full = get_filter_transmission_v2(filter_file_name2, filter_thickness2, 0.1)
        transmission_curve_full = transmission_curve1_full * transmission_curve2_full
    
    return energy_data, transmission_curve_full

def detector_response(energy):
    import numpy as np
    import pandas as pd
    K = np.zeros_like(energy)
    
    for i, e in enumerate(energy):
        if 10 <= e < 600:
            K[i] = 40 * (10)**(1/4) * np.exp(-e * np.log(10) / 480)
        elif 600 <= e <= 1800:
            K[i] = 10 * (10)**(1/2) * np.exp(-e * np.log(10) / 1200)
        else:
            K[i] = 1
    
    return K


def estimate_electron_temperature_2d_manual(shot_location, cal_location, shot_seq_port, parameters_location, where_to_look, pinhole_to_use1, pinhole_to_use2, where_to_place_max, filter_data_location, smoothing_type, smooth_pixels, bin_size, floor_value, how_to_bin, temp_bound, manual_or_not, flatfield):
    params = grab_params_v4(shot_seq_port, parameters_location)
    perp_range = real_units_to_pixels(0.3 / 2, params)
    
    S_total, did_switch, target_edge = pinhole_analysis_2d_manual(
        shot_location, cal_location, parameters_location, shot_seq_port,
        pinhole_to_use1[0], pinhole_to_use1[1], pinhole_to_use2[0], pinhole_to_use2[1],
        smoothing_type, smooth_pixels, where_to_look, np.flip(where_to_place_max), 
        perp_range, bin_size, floor_value, how_to_bin, params, manual_or_not, flatfield
    )
    S = S_total[:, :, 0]
    
    R = theoretical_analysis_2d(
        pinhole_to_use1[0], pinhole_to_use1[1], pinhole_to_use2[0], pinhole_to_use2[1],
        filter_data_location, did_switch, params
    )
    
    Te = np.zeros((S_total.shape[0], S_total.shape[1], 5))
    
    Te[:, :, 0] = calculate_temperature(R[1], S_total[:, :, 0], temp_bound)
    
    #Te[:, :, 1] = calculate_temperature(R[0], S_total[:, :, 1], temp_bound)
    #Te[:, :, 2] = calculate_temperature(R[2], S_total[:, :, 2], temp_bound)
    #Te[:, :, 3] = calculate_temperature(R[0], S_total[:, :, 3], temp_bound)
    #Te[:, :, 4] = calculate_temperature(R[2], S_total[:, :, 4], temp_bound)

        
    return Te, target_edge


def extract_average_temperature(temperature_map, where_to_start_box, window_dimensions, where_to_place_max, params, target_edge, color_bar_limit):
    import numpy as np
    import matplotlib.pyplot as plt

    # Define the conversion factors
    horizontal_conversion = temperature_map.shape[1] / 1051
    vertical_conversion = temperature_map.shape[0] / 1051
    
    # Calculate coordinates
    max_coords = [where_to_place_max[0] * horizontal_conversion, where_to_place_max[1] * vertical_conversion]
    zero_point = [max_coords[0], target_edge]
    
    # Calculate starting point and box size
    
    starting_point = np.zeros([2,1])
            
    starting_point[0] = np.round(real_units_to_pixels(where_to_start_box[0], params)).astype(int)
    starting_point[1] = np.round(real_units_to_pixels(where_to_start_box[1], params)).astype(int)

    starting_point = [int(horizontal_conversion * starting_point[0] + zero_point[0]), 
                      int(zero_point[1] - vertical_conversion * starting_point[1])]
        
    box_size = np.zeros([2,1])
    box_size[0] = np.round(real_units_to_pixels(window_dimensions[0], params)).astype(int)
    box_size[1] = np.round(real_units_to_pixels(window_dimensions[1], params)).astype(int)
    box_size = [int(horizontal_conversion * box_size[1]), int(vertical_conversion * box_size[0])]
    
    pixels_to_extract_x = np.arange(-box_size[0] // 2, box_size[0] // 2 + 1)
    pixels_to_extract_y = np.arange(-box_size[1] // 2, box_size[1] // 2 + 1)
    

    box = np.full((len(pixels_to_extract_x), len(pixels_to_extract_y)), np.nan)
    
    for i in range(0,len(pixels_to_extract_x)):
        for j in range(0,len(pixels_to_extract_y)):
        
            
            box[i, j] = temperature_map[starting_point[1] + pixels_to_extract_y[j], starting_point[0] + pixels_to_extract_x[i]]
    
    # Update temperature map with NaNs
    temperature_map[starting_point[1] + pixels_to_extract_y, starting_point[0] + pixels_to_extract_x.min()] = np.nan
    temperature_map[starting_point[1] + pixels_to_extract_y, starting_point[0] + pixels_to_extract_x.max()] = np.nan
    temperature_map[starting_point[1] + pixels_to_extract_y.min(), starting_point[0] + pixels_to_extract_x] = np.nan
    temperature_map[starting_point[1] + pixels_to_extract_y.max(), starting_point[0] + pixels_to_extract_x] = np.nan
    
    # Calculate temperature and error
    temperature = np.nanmean(box)
    error = np.nanstd(box, ddof=1)
    
    # Calculate length scales
    length_scale_x = pixels_to_real_units(np.arange(temperature_map.shape[1]) - where_to_place_max[0], params)
    length_scale_y = pixels_to_real_units(np.arange(temperature_map[:target_edge, :].shape[0]) - target_edge, params)
    
    # Plotting
    plt.figure()
    plt.imshow(temperature_map[:target_edge, :], extent=[length_scale_x.min(), length_scale_x.max(), length_scale_y.min(), length_scale_y.max()], origin='lower', cmap='hot')
    plt.colorbar(label="Temperature (eV)")
    plt.clim(100,color_bar_limit)
    plt.xlabel("Length [mm]", fontsize=15)
    plt.ylabel("Length [mm]", fontsize=15)
    plt.xlim(length_scale_x.min() - 0.05, length_scale_x.max() + 0.05)
    plt.ylim(length_scale_y.min() - 0.05, length_scale_y.max() + 0.05)
    plt.title("Temperature Map",)
    plt.grid(False)
    plt.show()
    
    return temperature, error


def extract_filter_file_name(params, row, column, filter_data_location):
    # Convert the (row, column) identifier method into the coordinate scheme of the excel sheet
    import numpy as np
    import pandas as pd
    value_to_select = (4 - row) * 4 + column - 1 #-1 is for python selection
            
    # Extract the filter material from the params file
    filter_materials = params['filter_mat'][0]

    filter_materials = filter_materials[value_to_select]    
    filter_material = filter_materials.split(',')
        
    if len(filter_material) == 1:  # Single filter material
        filter_file_name1 = f"{filter_data_location}{filter_material[0]}_10.txt"
        filter_file_name2 = 0
    elif len(filter_material) == 2:  # Two filter materials
        filter_file_name1 = f"{filter_data_location}{filter_material[0]}_10.txt"
        filter_file_name2 = f"{filter_data_location}{filter_material[1]}_10.txt"
    else:  # Safeguard against more than two filter materials
        raise ValueError("Too many filters. Readjust extract_filter_file_name.")
            
    return filter_file_name1, filter_file_name2


def extract_filter_thicknesses(params, row, column):
    # Convert the (row, column) identifier method into the coordinate scheme of the excel sheet
    import numpy as np
    import pandas as pd
    value_to_select = (4 - row) * 4 + column - 1 # -1 is for python
        
    # Extract the filter thicknesses from the params file
    filter_thicknesses = params['filter_thk']
    filter_thickness = filter_thicknesses.iloc[value_to_select]
    
    pinhole_filters = filter_thicknesses.iloc[value_to_select]
    
    filter_thickness1 = pinhole_filters[0]
    filter_thickness2 = pinhole_filters[1]
    
    if (np.isnan(filter_thickness2)):
        filter_thickness2 = 0
    
    return filter_thickness1, filter_thickness2


def filter_signal_ratio(signal_ratio):
    import numpy as np
    import pandas as pd
    # Filter out all the vacuum pixels from the signal ratio before calculating the electron temperature
    signal_ratio_filtered = np.where(signal_ratio > 1, np.nan, signal_ratio)
    return signal_ratio_filtered

def find_temp(ratio, signal_ratio, temps):
    # Find the temperature corresponding to the closest signal ratio
    index = np.argmin(np.abs(ratio - signal_ratio))
    return temps[index]

def floor_pinhole_2d(pinhole, floor_value):
    import numpy as np
    import pandas as pd
    # Floor the image by setting all signal values below floor_value to NaN
    pinhole_floored = np.where(pinhole < floor_value, np.nan, pinhole)
    return pinhole_floored

def get_filter_transmission_v2(file_name, new_thickness, error):
    import numpy as np
    import pandas as pd
    
    split_string = file_name.split('_')
    material_name = split_string[0]
    material_thickness = split_string[1].split('.')[0]
    standard_thickness = float(material_thickness.split('um')[0])
        
    # Read data from the file
    file_path = f"{material_name}_{int(standard_thickness)}um.txt"
    data = pd.read_csv(file_path, delim_whitespace=True, header=None).to_numpy()
    
    energy_data = data[2:, 0]
    transmission_coeff_data = data[2:, 1]
    
    transmission_coeff_data = transmission_coeff_data.astype(float)
    energy_data = energy_data.astype(float)
        
    if error == 0:
        T = np.power(transmission_coeff_data, (new_thickness / standard_thickness))
    else:
        upper_T = np.power(transmission_coeff_data,((new_thickness + error) / standard_thickness))
        lower_T = np.power(transmission_coeff_data,((new_thickness - error) / standard_thickness))
        new_trans_coeff = np.power(transmission_coeff_data,(new_thickness / standard_thickness))
        T = np.column_stack((upper_T, new_trans_coeff, lower_T))
    
    return energy_data, T


def grab_cal_params_v2(Date_Seq, parameters_location):
    import pandas as pd
    import numpy as np
    import ast

    # Suppress specific warnings
    pd.options.mode.chained_assignment = None
    
    # Define the file name and location
    file_name = 'cal_params_final.xlsx'
    file_location_and_name = f"{parameters_location}{file_name}"
    
    # Read in data and convert to a cell-like structure
    data = pd.read_excel(file_location_and_name)
    table = data.values.tolist()
    
    # Determine which column to use based on Date_Seq
    date_to_work_with = [i for i, x in enumerate(table[0]) if str(x) == Date_Seq]
    
    if not date_to_work_with:
        raise ValueError(f"No matching Date_Seq found for {Date_Seq}")
    
    date_to_work_with = date_to_work_with[0]
    
    # Construct the cal_params dictionary
    cal_params = {
        'date': float(table[1][date_to_work_with]),
        'seq': float(table[2][date_to_work_with]),
        'port': str(table[3][date_to_work_with]),
        'camera': str(table[4][date_to_work_with]),
        'strip_dir': str(table[5][date_to_work_with]),
        'volt': ast.literal_eval(table[6][date_to_work_with]),
        'dt': ast.literal_eval(table[7][date_to_work_with]),
        'vpulse': float(table[8][date_to_work_with]),
        'ccd_xwidth': float(table[9][date_to_work_with]),
        'ccd_ywidth': float(table[10][date_to_work_with]),
        'mirror': float(table[11][date_to_work_with])
    }
    
    return cal_params



def grab_params_v4(Date_Seq_Port, parameters_location):
    import pandas as pd
    import numpy as np
    import ast

    # Suppress specific warnings
    pd.options.mode.chained_assignment = None
    
    # Define the file name and location
    file_name = 'xray_params_final_python.xlsx'
    file_location_and_name = f"{parameters_location}{file_name}"
    
    # Read in data and convert to a cell-like structure
    data = pd.read_excel(file_location_and_name)
    table = data.values.tolist()
        
    # Determine which column to use based on Date_Seq_Port
    date_to_work_with = [i for i, x in enumerate(table[0]) if str(x) == Date_Seq_Port]
    
    if not date_to_work_with:
        raise ValueError(f"No matching Date_Seq_Port found for {Date_Seq_Port}")
    
    date_to_work_with = date_to_work_with[0]
    
    
    # Construct the params dictionary
    params = {
        'expr': str(table[0][date_to_work_with]),
        'date': float(table[1][date_to_work_with]),
        'seq': float(table[2][date_to_work_with]),
        'port': str(table[3][date_to_work_with]),
        'camera': str(table[4][date_to_work_with]),
        'ints': float(table[5][date_to_work_with]),
        'cal_cam': str(table[6][date_to_work_with]),
        'cal_shot': str(table[7][date_to_work_with]),
        'strip_dir': str(table[8][date_to_work_with]),
        't0': float(table[9][date_to_work_with]),
        'mag': float(table[10][date_to_work_with]),
        'dt': ast.literal_eval(table[11][date_to_work_with]),
        'vpulse': float(table[12][date_to_work_with]),
        'ccd_xwidth': float(table[13][date_to_work_with]),
        'ccd_ywidth': float(table[14][date_to_work_with]),
        'ang': float(table[15][date_to_work_with]),
        'mirror': float(table[16][date_to_work_with]),
        'npinholes': float(table[17][date_to_work_with]),
        'nrow': float(table[18][date_to_work_with]),
        'ncol': float(table[19][date_to_work_with])
    }
    
    # Extract pinholes data
    index = 21
    pinholes = []
    while not np.isnan(float(table[index][date_to_work_with])):
        pinholes.append(float(table[index][date_to_work_with]))
        index += 1
    params['pinholes'] = pd.DataFrame(pinholes)
    
    # Extract filter materials data
    index = 38
    filter_materials = []
    while (str(table[index][date_to_work_with]) != "nan"):
        filter_materials.append(str(table[index][date_to_work_with]))
        index += 1
        
    params['filter_mat'] = pd.DataFrame(filter_materials)
    
    # Extract filter thicknesses data
    filter_thicknesses = []
    
    for index in range(59, len(data)):
                
        if isinstance(table[index][date_to_work_with],str):
            filter_thicknesses.append(ast.literal_eval(table[index][date_to_work_with]))
            
        elif isinstance(table[index][date_to_work_with],float):
            filter_thicknesses.append(table[index][date_to_work_with])

        
    params['filter_thk'] = pd.DataFrame(filter_thicknesses)
        
    return params

def horizontally_align_pinholes_2D_v2(inputPinhole1, inputPinhole2, whereToLook, horzMax, vertMax, stuck, manualShift):
    # This function will take in two pinholes and horizontally align pinhole 2 wrt to Pinhole 1.
    import numpy as np
    horizontalLineout1 = np.nanmean(inputPinhole1[vertMax-30:vertMax+30, :], axis=0)  # Generating lineout through pinhole 1
    horizontalLineout2 = np.nanmean(inputPinhole2[vertMax-30:vertMax+30, :], axis=0)  # Generating lineout through pinhole 2

    # Depending on the feature we are interested in, the following selection statements will identify the proper region to align to.
    if whereToLook in [1, -1]:
        grad_lineout1 = np.gradient(horizontalLineout1[horzMax:])
        grad_lineout2 = np.gradient(horizontalLineout2[horzMax:])

        alignmentPoint1 = np.argmin(grad_lineout1)  # Isolating alignment point in pinhole 1
        alignmentPoint2 = np.argmin(grad_lineout2)  # Isolating point to align in pinhole 2

    elif whereToLook == 0:
        grad_lineout1 = np.gradient(horizontalLineout1[horzMax-25:horzMax+25])
        grad_lineout2 = np.gradient(horizontalLineout2[horzMax-25:horzMax+25])

        alignmentPoint1 = np.argmin(np.abs(grad_lineout1))  # Isolating alignment point in pinhole 1
        alignmentPoint2 = np.argmin(np.abs(grad_lineout2))  # Isolating point to align in pinhole 2

    if stuck == 0:
        difference = alignmentPoint1 - alignmentPoint2  # Calculating difference in alignment points
    elif stuck == 1:
        difference = manualShift

    imbedLength = 500  # How much blank space do we want on both sides of the array? Blank space on each side = imbedLength/2
    imbededSpace = np.full((inputPinhole2.shape[0] + imbedLength, inputPinhole2.shape[1] + imbedLength), np.nan)  # Creating empty NaN space
    rangeX = np.arange(imbedLength//2, imbedLength//2 + inputPinhole2.shape[1])  # Creating range for the horizontal values in the data
    rangeY = np.arange(inputPinhole2.shape[0])  # Creating range for the vertical values in the data
    imbededSpace[rangeY, rangeX] = inputPinhole2  # Imbedding pinhole2 in the empty NaN Space

    alignedPinhole2_h = imbededSpace[rangeY, rangeX - difference]  # Applying shift to pinhole 2, aligning it wrt pinhole 1, and outputting the new aligned pinhole 2

    return alignedPinhole2_h, difference


def id_pinhole_maxima(pinhole, whereToLook):
    import numpy as np
    # This function locates the maxima of a given pinhole image of arbitrary size

    length, width = pinhole.shape

    if whereToLook in [-1, 1]:
        xCoord = np.argmax(np.max(pinhole[length//2-10:length//2+10, width//2-10:width//2+25], axis=0))
        xCoord = round(xCoord + width//2 - 10)

        yCoord = np.argmax(pinhole[:, xCoord])

        coords = [yCoord, xCoord]

    elif whereToLook == 0:
        xCoord = np.argmax(np.max(pinhole[length//2-25:length//2+25, width//2-50:width//2+50], axis=0))
        xCoord = round(xCoord + width//2 - 50)

        yCoord = np.argmax(pinhole[:, xCoord])

        coords = [yCoord, xCoord]

    return coords


def isolate_region_of_interest(lineout, lineout2, error_pixels, Range):
    import numpy as np
    # Process w/o error
    new_max = 500
    bckgrnd = background(lineout)  # Calculating the Background
    lineout_n = lineout[new_max::-1][new_max-Range:]  # Isolating the Region of Interest
    lineout_n = lineout_n - bckgrnd  # Background subtraction
    lineout_n = np.maximum(lineout_n, 0)  # Flooring the < 0 values
    norm = np.max(lineout_n)  # Identifying the value of the max
    lineout_n = (1/norm) * lineout_n  # Normalizing 

    bckgrnd = background(lineout2)
    lineout2_n = lineout2[new_max::-1][new_max-Range:]
    lineout2_n = lineout2_n - bckgrnd
    lineout2_n = np.maximum(lineout2_n, 0)
    lineout2_n = (1/norm) * lineout2_n

    # Upper error bound calculation
    new_max = 500 + error_pixels
    lineout_u = lineout[new_max::-1][new_max-Range:]
    lineout_u = lineout_u - bckgrnd
    lineout_u = np.maximum(lineout_u, 0)
    norm = np.max(lineout_u)
    lineout_u = (1/norm) * lineout_u

    lineout2_u = lineout2[new_max::-1][new_max-Range:]
    lineout2_u = lineout2_u - bckgrnd
    lineout2_u = np.maximum(lineout2_u, 0)
    lineout2_u = (1/norm) * lineout2_u

    # Lower Bound
    new_max = 500 - error_pixels
    lineout_L = lineout[new_max::-1][new_max-Range:]
    lineout_L = lineout_L - bckgrnd
    lineout_L = np.maximum(lineout_L, 0)
    norm = np.max(lineout_L)
    lineout_L = (1/norm) * lineout_L

    lineout2_L = lineout2[new_max::-1][new_max-Range:]
    lineout2_L = lineout2_L - bckgrnd
    lineout2_L = np.maximum(lineout2_L, 0)
    lineout2_L = (1/norm) * lineout2_L

    full_lineout = np.concatenate([lineout_u, lineout_n, lineout_L])
    full_lineout2 = np.concatenate([lineout2_u, lineout2_n, lineout2_L])

    return full_lineout, full_lineout2


def is_stuck(arrayOfShifts):
    import numpy as np
    # This function simply accepts an array of shifts produced by a given iteration of alignment and attempts to determine if the alignment algorithm is stuck

    if len(arrayOfShifts) < 20:
        # Let it go for 20 iterations and then check
        return 0, 0

    sumOfPreviousShifts = sum(arrayOfShifts[-3:])

    if sumOfPreviousShifts <= 20:  # Way of seeing if it's circling around the correct answer
        previousIterations = arrayOfShifts[-3:]

        manualShift = np.prod(np.abs(previousIterations))**(1/len(previousIterations))  # Geometric mean
        manualShift = round(manualShift / 2)

        if arrayOfShifts[-1] > 0:
            manualShift = -1 * manualShift

        return 1, manualShift

    elif sumOfPreviousShifts > 20:  # Way of seeing if it's circling around the correct answer
        return 2, 0

def iterative_pinhole_alignment_2d(input_pinhole1, input_pinhole2, where_to_look, average_range, maxima_coords, stuck, manual_shift):
    input_pinhole2_v, shift_vert, target_edge = vertically_align_pinholes_2d_v2(input_pinhole1, input_pinhole2, average_range, maxima_coords[0], maxima_coords[1], stuck[0], manual_shift[0])
    input_pinhole2_hv, shift_horz = horizontally_align_pinholes_2d_v2(input_pinhole1, input_pinhole2_v, where_to_look, maxima_coords[1], maxima_coords[0], stuck[1], manual_shift[1])

    import numpy as np
    import matplotlib.pyplot as plt
    
    pinhole1_aligned = input_pinhole1
    pinhole2_aligned = input_pinhole2_hv

    did_shift_vert = (shift_horz != 0)
    did_shift_horz = (shift_vert != 0)
    keep_going = (did_shift_vert or did_shift_horz)
    keep_going = (stuck[0] != 1 and stuck[1] != 1)

    return pinhole1_aligned, pinhole2_aligned, keep_going, shift_horz, shift_vert, target_edge

def make_processed_data_v2(shot_location, cal_location, parameters_location, shot_seq_port, flatfield):
    params = grab_params_v4(shot_seq_port, parameters_location)
    cal_params = grab_cal_params_v2(params['cal_shot'], parameters_location)
    shot_data, cal_data = read_hdf5_data(shot_location, cal_location)
    
    import numpy as np
    import matplotlib.pyplot as plt

    ang = (2 * np.pi / 360) * params['ang']
    rot = np.array([[np.cos(ang), -np.sin(ang)], [np.sin(ang), np.cos(ang)]])
    shot_data = mirror(shot_data, params['mirror'])
    cal_data = mirror(cal_data, cal_params['mirror'])
    avg = np.mean(cal_data)
    flatfield = cal_data / avg

    flatfield[flatfield == 0] = 1
    raw_data = shot_data / flatfield

    return raw_data

def manual_pinhole_alignment(pinhole1, pinhole2, perp_range, where_to_place_max):
    
    import numpy as np
    import matplotlib.pyplot as plt
    
    keep_going = 1
    shift_x = 0
    shift_y = 0
    pinhole1_to_work_with = np.copy(pinhole1)
    pinhole2_to_work_with = np.copy(pinhole2)
    imbed_length = 200
    imbeded_space = np.full((pinhole2.shape[0] + imbed_length, pinhole2.shape[1] + imbed_length), np.nan)
    range_y = slice(imbed_length // 2, imbed_length // 2 + pinhole2.shape[0])
    range_x = slice(imbed_length // 2, imbed_length // 2 + pinhole2.shape[1])
    imbeded_space[range_y, range_x] = pinhole2_to_work_with
    target_edge = target_edge_finder(pinhole1, where_to_place_max[0], where_to_place_max[1], perp_range)

    vertical_region_to_draw = slice(where_to_place_max[1] - 2, where_to_place_max[1] + 3)
    horz_region_to_draw = slice(where_to_place_max[0] - 2, where_to_place_max[0] + 3)
    vertical_lineout1 = np.nanmean(pinhole1_to_work_with[:, vertical_region_to_draw], axis=1)
    horz_lineout1 = np.nanmean(pinhole1_to_work_with[horz_region_to_draw, :], axis=0)

    pinhole1_to_work_with[:, vertical_region_to_draw] = np.nan
    pinhole1_to_work_with[horz_region_to_draw, :] = np.nan
    pinhole1_to_work_with[:target_edge, :] = np.nan

    fig, ax = plt.subplots(2, 3, figsize=(20, 10))

    ax[0, 0].imshow(pinhole1_to_work_with, aspect='auto', vmin=0, vmax=1200)
    ax[0, 0].set_title('Pinhole 1 - Fixed')
    ax[0, 0].set_xlabel('CCD Width (Pixels)')
    ax[0, 0].set_ylabel('CCD Length (Pixels)')
    ax[0, 0].grid(True)

    ax[0, 1].plot(vertical_lineout1, 'b', linewidth=2.5)
    ax[0, 1].axvline(where_to_place_max[0], color='k', linewidth=2.5)
    ax[0, 1].set_title('Vertical Lineout Down Maxima of Pinhole 1')
    ax[0, 1].set_xlabel('CCD Length (Pixels)')
    ax[0, 1].set_ylabel('Counts (Arb. Units)')

    ax[0, 2].plot(horz_lineout1, 'r', linewidth=2.5)
    ax[0, 2].axvline(where_to_place_max[1], color='k', linewidth=2.5)
    ax[0, 2].set_title('Horizontal Lineout Through Maxima of Pinhole 1')
    ax[0, 2].set_xlabel('CCD Width (Pixels)')
    ax[0, 2].set_ylabel('Counts (Arb. Units)')

    while keep_going == 1:
        pinhole2_to_work_with = imbeded_space[range_y.start - shift_y:range_y.stop - shift_y, range_x.start - shift_x:range_x.stop - shift_x]
        vertical_lineout2 = np.nanmean(pinhole2_to_work_with[:, vertical_region_to_draw], axis=1)
        horz_lineout2 = np.nanmean(pinhole2_to_work_with[horz_region_to_draw, :], axis=0)

        pinhole2_for_plotting = np.copy(pinhole2_to_work_with)
        pinhole2_for_plotting[:, vertical_region_to_draw] = np.nan
        pinhole2_for_plotting[horz_region_to_draw, :] = np.nan
        pinhole2_for_plotting[:target_edge, :] = np.nan
        
        ax[1, 0].imshow(pinhole2_for_plotting, aspect='auto', vmin=0, vmax=500)
        ax[1, 0].set_title('Pinhole 2 - Mutable')
        ax[1, 0].set_xlabel('CCD Width (Pixels)')
        ax[1, 0].set_ylabel('CCD Length (Pixels)')
        ax[1, 0].grid(True)

        ax[1, 1].plot(vertical_lineout2, 'b', linewidth=2.5)
        ax[1, 1].axvline(where_to_place_max[0], color='k', linewidth=2.5)
        ax[1, 1].set_title('Vertical Lineout of Pinhole 2 Down Maxima of Pinhole 1')
        ax[1, 1].set_xlabel('CCD Length (Pixels)')
        ax[1, 1].set_ylabel('Counts (Arb. Units)')

        ax[1, 2].plot(horz_lineout2, 'r', linewidth=2.5)
        ax[1, 2].axvline(where_to_place_max[1], color='k', linewidth=2.5)
        ax[1, 2].set_title('Horizontal Lineout of Pinhole 2 Through Maxima of Pinhole 1')
        ax[1, 2].set_xlabel('CCD Width (Pixels)')
        ax[1, 2].set_ylabel('Counts (Arb. Units)')

        plt.draw()

        keep_going = int(input("Would you like to adjust Pinhole 2? (1 == Yes, 0 == No): "))

        if keep_going == 1:
            new_shift_x = int(input("Please enter the HORIZONTAL adjustment to Pinhole 2 in pixels ([+] integers move the image RIGHT, [-] integers move the image LEFT): "))
            new_shift_y = int(input("Please enter the VERTICAL adjustment to Pinhole 2 in pixels ([+] integers move the image UP, [-] integers move the image DOWN): "))

            shift_x += new_shift_x
            shift_y -= new_shift_y

    print("Alignment Done!")
    aligned_pinhole2 = imbeded_space[range_y.start - shift_y:range_y.stop - shift_y, range_x.start - shift_x:range_x.stop - shift_x]

    return aligned_pinhole2, target_edge


def mirror(data, flip_number):
    import numpy as np

    if flip_number == 1:
        # Y axis
        shot_data = np.flip(data, axis=0)
        shot_data = np.rot90(shot_data)

    elif flip_number == 2:
        # X axis and Y Axis
        shot_data = np.flip(data, axis=1)
        shot_data = np.flip(shot_data, axis=0)
        shot_data = np.rot90(shot_data)

    elif flip_number == 3:
        # Just X Axis
        shot_data = np.flip(data, axis=1)
        shot_data = np.rot90(shot_data)

    return shot_data


def normalize_v2(lineout, where_to_look):
    import numpy as np

    first_nan_index = np.isnan(lineout).argmin()
    section_constant = round(len(lineout) / 3)
    
    if where_to_look == -1:
        norm_range = slice(first_nan_index, section_constant)
    elif where_to_look == 0:
        norm_range = slice(section_constant, 2 * section_constant)
    elif where_to_look == 1:
        norm_range = slice(2 * section_constant, 3 * section_constant)
    
    norm = np.nanmax(lineout[norm_range])
    normalized_lineout = lineout / norm
    return normalized_lineout


def pinhole_analysis_2d_manual(shot_location, cal_location, parameters_location, shot_seq_port, row1, col1, row2, col2, 
                               smoothing_type, smooth_pixels, where_to_look, where_to_place_max, perp_range, bin_size, 
                               floor_value, how_to_bin, params, manual_or_not, flatfield):
    import numpy as np

    
    # Creating Raw Data + Isolating Pinholes
    raw_data = make_processed_data_v2(shot_location, cal_location, parameters_location, shot_seq_port, flatfield)
    
    pinhole1,p1_max_x,p1_max_y = select_pinhole_2d(raw_data, row1, col1, where_to_place_max[0], where_to_place_max[1], 
                                 smoothing_type, smooth_pixels, where_to_look, 1)
    pinhole2,p2_max_x,p2_max_y = select_pinhole_2d(raw_data, row2, col2, where_to_place_max[0], where_to_place_max[1], 
                                 smoothing_type, smooth_pixels, where_to_look, 1)
      
    pinhole1, pinhole2, did_switch = switch_pinholes(pinhole1, pinhole2)
    
    # Aligning the Pinholes
    if manual_or_not == 0:  # Automated Alignment Algorithm
        pinhole2_aligned, target_edge = align_pinholes_2d_v2(pinhole1, pinhole2, where_to_look, perp_range, where_to_place_max)
        pinhole1_aligned = pinhole1
    elif manual_or_not == 1:  # Manual Alignment
        pinhole2_aligned, target_edge = manual_pinhole_alignment(pinhole1, pinhole2, perp_range, where_to_place_max)
        pinhole1_aligned = pinhole1
    
    pinhole1_final = process_pinholes(pinhole1_aligned, where_to_place_max, floor_value)
    pinhole2_final = process_pinholes(pinhole2_aligned, where_to_place_max, floor_value)
    
    pinhole1_final[:target_edge,:] = np.nan
    pinhole2_final[:target_edge,:] = np.nan

    length_scale_x = pixels_to_real_units(np.arange(len(pinhole1_final)) - where_to_place_max[1], params)
    length_scale_y = pixels_to_real_units(np.arange(target_edge) - target_edge, params)
    
    plotArray(pinhole1_final)
    plotArray(pinhole2_final)

    # Processing Pinholes + Constructing the S matrices
    s_initial = pinhole2_final / pinhole1_final
    #s = binning_algorithm_v2(s_initial, bin_size[0], bin_size[1], how_to_bin)
    s = s_initial
    
    
    error_matrix = np.full((pinhole2_final.shape[0] + 500, pinhole2_final.shape[1] + 500), np.nan)
    imbed_range = np.arange(pinhole2_final.shape[0]) + 250
    
    error_matrix[imbed_range[:, None], imbed_range] = pinhole2_final
    
    error_shift = round(real_units_to_pixels(0.05, params))
    
    upper_error_bound_h = error_matrix[imbed_range[:, None], imbed_range + error_shift]
    lower_error_bound_h = error_matrix[imbed_range[:, None], imbed_range - error_shift]
    
    upper_error_bound_v = error_matrix[imbed_range + error_shift, imbed_range[:, None]]
    lower_error_bound_v = error_matrix[imbed_range - error_shift, imbed_range[:, None]]
    
    s_horizontal_upper_bound = upper_error_bound_h / pinhole1_final
    s_horizontal_lower_bound = lower_error_bound_h / pinhole1_final
    
    s_vertical_upper_bound = upper_error_bound_v / pinhole1_final
    s_vertical_lower_bound = lower_error_bound_v / pinhole1_final
    
    s_horizontal_upper_bound = binning_algorithm_v2(s_horizontal_upper_bound, bin_size[0], bin_size[1], how_to_bin)
    s_horizontal_lower_bound = binning_algorithm_v2(s_horizontal_lower_bound, bin_size[0], bin_size[1], how_to_bin)
    
    s_vertical_upper_bound = binning_algorithm_v2(s_vertical_upper_bound, bin_size[0], bin_size[1], how_to_bin)
    s_vertical_lower_bound = binning_algorithm_v2(s_vertical_lower_bound, bin_size[0], bin_size[1], how_to_bin)
    
    s_total = np.zeros((s.shape[0], s.shape[1], 5))
    s_total[:, :, 0] = s
    s_total[:, :, 1] = s_horizontal_upper_bound
    s_total[:, :, 2] = s_horizontal_lower_bound
    s_total[:, :, 3] = s_vertical_upper_bound
    s_total[:, :, 4] = s_vertical_lower_bound
    
    import matplotlib.pyplot as plt
    
    plt.figure()
    plt.imshow(s, cmap='viridis', aspect='auto')
    plt.colorbar()
    plt.clim(0, 1)  # Set color bar limits
    #plt.xlim([0, 4200])
    #plt.ylim([0, 4200])
    plt.title('Signal Ratio')
    plt.grid(False)
    plt.show()
    
    target_edge = round(target_edge * (s.shape[0] / pinhole1.shape[1]))
    
    length_scale_x = pixels_to_real_units(np.arange(len(s)), params)
    length_scale_y = pixels_to_real_units(np.arange(target_edge), params)
    
    return s_total, did_switch, target_edge


def pixels_to_real_units(pixels, params):
    import numpy as np
    import matplotlib.pyplot as plt
    from scipy.ndimage import gaussian_filter
    
    return (36/4) * (1/1050) * (1/params['mag']) * pixels

def plotArray(data):
    
    import matplotlib.pyplot as plt
    
    plt.figure()
    plt.imshow(data, cmap='viridis', aspect='auto')
    plt.colorbar()
    plt.clim(0, 1200)  # Set color bar limits
    #plt.xlim([0, 4200])
    #plt.ylim([0, 4200])
    plt.grid(False)
    plt.show()

def plot_pinhole(data):
    
    import matplotlib.pyplot as plt
    
    plt.figure()
    plt.imshow(data, cmap='viridis', aspect='auto')
    plt.colorbar()
    plt.clim(0, 1200)  # Set color bar limits
    plt.xlim([0, data.shape[1]])
    plt.ylim([0, data.shape[0]])
    plt.grid(False)
    plt.xlabel('CCD Width (Pixels)', fontweight='bold')
    plt.ylabel('CCD Length (Pixels)', fontweight='bold')
    plt.show()

def plotTempMap(tempLength, Te, coloBarLimit):
    
    import matplotlib.pyplot as plt

    plt.figure()
    plt.imshow(Te, cmap='hot', aspect='auto')
    plt.colorbar(label="Temperature (eV)")
    plt.xlabel("Length [mm]", fontsize=24)
    plt.ylabel("Length [mm]", fontsize=24)
    plt.clim(0, coloBarLimit)  # Set color bar limits
    plt.gca().set_linewidth(2)
    plt.gca().tick_params(labelsize=24)
    plt.grid(False)
    plt.title("Temperature Map")
    plt.show()

def process_pinholes(inputPinhole, maximaCoords, floorValue):
    import numpy as np
    import matplotlib.pyplot as plt

    background = background_2d(inputPinhole, maximaCoords)  # Define the background_2D function in Python
    pinhole_b = inputPinhole - background
                
    pinhole_bf = floor_pinhole_2d(pinhole_b, floorValue)

    return pinhole_bf

def read_hdf5_data(shot_location, cal_location):
    import h5py
    import numpy as np

    # Read in the pre-shot data
    with h5py.File(shot_location, 'r') as f:
        preshot = f['/ATTRIBUTES/PRESHOT_IMAGESPRESHOT_IMAGES/0/DATA/DATA/'][:]
        shot = f['/ATTRIBUTES/SHOT_IMAGE/DATA/DATA/'][:]
    
    # Convert data types to int32
    preshot = preshot.astype(np.int32)
    shot = shot.astype(np.int32)
    
    # Background subtraction
    shot_data = shot - preshot
    shot_data[shot_data < 0] = 0  # Set negative values to 0

    # Read in calibration preshot and shot data
    with h5py.File(cal_location, 'r') as f:
        cal_preshot = f['/ATTRIBUTES/PRESHOT_IMAGESPRESHOT_IMAGES/0/DATA/DATA/'][:]
        cal_shot = f['/ATTRIBUTES/SHOT_IMAGE/DATA/DATA/'][:]
    
    # Convert calibration data types to int32
    cal_preshot = cal_preshot.astype(np.int32)
    cal_shot = cal_shot.astype(np.int32)
    
    # Background subtraction
    cal_data = cal_shot - cal_preshot
    cal_data[cal_data < 0] = 0  # Set negative values to 0

    # Convert to single precision (float32)
    shot_data = shot_data.astype(float)
    cal_data = cal_data.astype(float)
    
    return shot_data, cal_data


def real_units_to_pixels(x, params):

    return (1050 * (4/36) * x * params['mag'])


def select_pinhole_2d(RawData, row, col, max_row, max_col, smooth_type, smooth_pixels, location, smoothing_on_off):
    import numpy as np
    import matplotlib.pyplot as plt
    from scipy.ndimage import gaussian_filter
   
    Master_array = np.full((6200, 6200), np.nan, dtype=float)
    Size_x, Size_y = RawData.shape
    Master_array[1000:(Size_x+1000), 1000:(Size_y+1000)] = RawData

    n = Size_x / 4
    m = Size_y / 4

    x = np.arange((col-1)*m, col*m+1, 1).astype(int)
    y = np.arange((row-1)*n, row*n+1, 1).astype(int)

    if x.min() == 0:
        x += 1
    if y.min() == 0:
        y += 1

    max_x_to_smooth = x.max() + 300
    max_y_to_smooth = y.max() + 300

    min_x_to_smooth = x.min() - 300
    min_y_to_smooth = y.min() - 300

    x_smoothing_range = np.arange(min_x_to_smooth, max_x_to_smooth+1, 1).astype(int)
    y_smoothing_range = np.arange(min_y_to_smooth, max_y_to_smooth+1, 1).astype(int)
    
    
    Array_to_smooth = Master_array[(min_y_to_smooth+999):(max_y_to_smooth+999), (min_x_to_smooth+999):(max_x_to_smooth+999)]
    #Array_to_smooth = Master_array[y_smoothing_range+999, x_smoothing_range+999]

    if smooth_type == "mean":
        Smoothed_array = np.apply_along_axis(lambda m: np.convolve(m, np.ones(smooth_pixels)/smooth_pixels, mode='same'), axis=0, arr=Array_to_smooth)
        Smoothed_array = np.apply_along_axis(lambda m: np.convolve(m, np.ones(smooth_pixels)/smooth_pixels, mode='same'), axis=1, arr=Smoothed_array)
    elif smooth_type == "gauss":
        Smoothed_array = gaussian_filter(Array_to_smooth, smooth_pixels)

    Unsmoothed_data = Master_array

    
    Master_array[(min_y_to_smooth+999):(max_y_to_smooth+999), (min_x_to_smooth+999):(max_x_to_smooth+999)] = Smoothed_array
    Smoothed_master_array = Master_array
    

    Guess_region_1 = Smoothed_master_array[(y.min()+999):(y.max()+999), (x.min()+999):(x.max()+999)] #should be 999


    if location == -1:
        guess_lower_bound_in_y = 250
        guess_lower_bound_in_x = 125
    elif location == 0:
        guess_lower_bound_in_x = 400
        guess_lower_bound_in_y = 50
    elif location == 1:
        guess_lower_bound_in_x = 500
        guess_lower_bound_in_y = 250

    Guess_region_2 = Guess_region_1[guess_lower_bound_in_y:800, guess_lower_bound_in_x:900]
    old_y_max_location_in_guess_2, old_x_max_location_in_guess_2 = np.unravel_index(np.argmax(Guess_region_2, axis=None), Guess_region_2.shape)
    old_y_max_location = old_y_max_location_in_guess_2 + guess_lower_bound_in_y
    old_x_max_location = old_x_max_location_in_guess_2 + guess_lower_bound_in_x

    diff_x = max_col - old_x_max_location
    diff_y = max_row - old_y_max_location

    #Corrected_x_range = x - diff_x + 999
    #Corrected_y_range = y - diff_y + 999

    if smoothing_on_off == 1:
        Final = Smoothed_master_array[(y.min()-diff_y+999):(y.max()-diff_y+999), (x.min()-diff_x+999):(x.max()-diff_x+999)]
    else:
        Final = Unsmoothed_data[(y.min()-diff_y+999):(y.max()-diff_y+999), (x.min()-diff_x+999):(x.max()-diff_x+999)]

    Guess_region_2 = Final[475:525, 475:525]
    new_y_max_location, new_x_max_location = np.unravel_index(np.argmax(Guess_region_2, axis=None), Guess_region_2.shape)
    new_y_max_location += 474
    new_x_max_location += 474
       
    return Final, new_x_max_location, new_y_max_location
        
def switch_pinholes(pinhole1, pinhole2):
    import numpy as np

    #This function is used to flip the data, as the pipeline assumes 
    #that the left pinhole in the isolated pair is the high-signal, low-filtered pinhole,
    #and the right pinhole is the low-signal, high-filtered pinhole.
            
    if np.mean(pinhole2[474:525, 474:575]) > np.mean(pinhole1[474:525, 474:575]):
        didSwitch = 1
        pinhole1_new = pinhole2
        pinhole2_new = pinhole1
    else:
        didSwitch = 0
        pinhole1_new = pinhole1
        pinhole2_new = pinhole2

    return pinhole1_new, pinhole2_new, didSwitch

    
def gaussian(x, a, b, c):
    return a * np.exp(-((x - b) ** 2) / (2 * c ** 2))


def target_edge_finder(inputPinhole, vertMax, horzMax, averageRange):

    #This function is designed for locating the edge of a given target.
    from scipy.optimize import curve_fit
    import numpy as np
    import matplotlib.pyplot as plt

    lineout = np.mean(inputPinhole[:, (horzMax-round(averageRange)):(horzMax+round(averageRange))], axis=1)
    lineoutToFit = lineout
   
    regionToFit = lineoutToFit[:vertMax - 10]

    pixelSpace = np.arange(1, len(regionToFit) + 1)
        
    whereAreNans = np.isnan(regionToFit)
    whereAreNonNans = [not elem for elem in whereAreNans]

    # Fit the region with a Gaussian function
    popt, _ = curve_fit(gaussian, pixelSpace[whereAreNonNans], regionToFit[whereAreNonNans])

    # Calculate the maximum gradient to find the alignment point
    gradient_values = np.gradient(gaussian(pixelSpace, *popt))
    alignmentPoint = np.argmax(gradient_values)
    
    targetEdge = alignmentPoint

    return targetEdge


def theoretical_analysis_2d(row1, col1, row2, col2, filterDataLocation, didSwitch, params):
    import numpy as np


    #This function performs theoretical analysis based on pinhole filter data.

    if didSwitch == 0:
        # Extracting filter information if we did not switch the pinholes
        filterFileName1_pinhole1, filterFileName2_pinhole1 = extract_filter_file_name(params, row1, col1, filterDataLocation)
        filterFileName1_pinhole2, filterFileName2_pinhole2 = extract_filter_file_name(params, row2, col2, filterDataLocation)

        thickness1_pinhole1, thickness2_pinhole1 = extract_filter_thicknesses(params, row1, col1)
        thickness1_pinhole2, thickness2_pinhole2 = extract_filter_thicknesses(params, row2, col2)
    else:
        # Extracting filter information if we did switch the pinholes
        filterFileName1_pinhole1, filterFileName2_pinhole1 = extract_filter_file_name(params, row1, col2, filterDataLocation)
        filterFileName1_pinhole2, filterFileName2_pinhole2 = extract_filter_file_name(params, row2, col1, filterDataLocation)

        thickness1_pinhole1, thickness2_pinhole1 = extract_filter_thicknesses(params, row2, col2)
        thickness1_pinhole2, thickness2_pinhole2 = extract_filter_thicknesses(params, row1, col1)

    # Calculating the transmission curves for both pinhole filters
    energyData, transmissionCurve1_full = calculate_transmission_curves(filterFileName1_pinhole1, filterFileName2_pinhole1, thickness1_pinhole1, thickness2_pinhole1)
    energyData, transmissionCurve2_full = calculate_transmission_curves(filterFileName1_pinhole2, filterFileName2_pinhole2, thickness1_pinhole2, thickness2_pinhole2)
    
    # Defining physical parameters
    temperatureSpace = np.logspace(2, 4, 1001)

    # Integration
    Integral1_upperBound = calculate_theoretical_ratio(energyData, temperatureSpace, transmissionCurve1_full[:, 0])
    Integral2_upperBound = calculate_theoretical_ratio(energyData, temperatureSpace, transmissionCurve2_full[:, 0])

    Integral1 = calculate_theoretical_ratio(energyData, temperatureSpace, transmissionCurve1_full[:, 1])
    Integral2 = calculate_theoretical_ratio(energyData, temperatureSpace, transmissionCurve2_full[:, 1])

    Integral1_lowerBound = calculate_theoretical_ratio(energyData, temperatureSpace, transmissionCurve1_full[:, 2])
    Integral2_lowerBound = calculate_theoretical_ratio(energyData, temperatureSpace, transmissionCurve2_full[:, 2])

    G_1 = Integral2_upperBound / Integral1_lowerBound
    G_2 = Integral2_lowerBound / Integral1_upperBound
    R_N = Integral2 / Integral1

    R = [G_1, R_N, G_2]

    return R
        
def vertically_align_pinholes_2d_v2(inputPinhole1, inputPinhole2, averageRange, vertMax, horzMax, stuck, manualShift):
    #This function takes in two pinholes and vertically aligns pinhole 2 with respect to pinhole 1.

    lineout1 = np.nanmean(inputPinhole1[:, (horzMax-averageRange):(horzMax+averageRange)], axis=1)
    lineout2 = np.nanmean(inputPinhole2[:, (horzMax-averageRange):(horzMax+averageRange)], axis=1)

    # Flip the lineouts for easier gradient calculations
    lineoutToFit1 = np.flip(lineout1)
    lineoutToFit2 = np.flip(lineout2)
    
    regionToFit1 = lineoutToFit1[:vertMax + 50]
    regionToFit2 = lineoutToFit2[:vertMax + 50]
    
    pixelSpace = np.arange(1, len(regionToFit1) + 1)
    
    nanIndices = np.isnan(regionToFit2)
    
    fit1 = curve_fit(gaussian, pixelSpace, regionToFit1)
    fit2 = curve_fit(gaussian, pixelSpace[~nanIndices], regionToFit2[~nanIndices])
    
    # Calculate alignment points
    alignmentPoint1 = np.argmax(np.gradient(gaussian(pixelSpace, *fit1[0])))
    alignmentPoint2 = np.argmax(np.gradient(gaussian(pixelSpace, *fit2[0])))

    targetEdge = len(lineout1) - alignmentPoint1

    imbedLength = 500
    
    # Creating embedded space
    imbededSpace = np.full((inputPinhole2.shape[0] + imbedLength, inputPinhole2.shape[1]), np.nan)
    rangeY = np.arange(imbedLength // 2, (imbedLength // 2) + inputPinhole2.shape[0])
    rangeX = np.arange(inputPinhole2.shape[1])
    
    imbededSpace[rangeY[:, None], rangeX] = inputPinhole2

    if stuck == 0:
        difference = -1 * (alignmentPoint1 - alignmentPoint2)
    else:
        difference = manualShift

    alignedPinhole2_hv = imbededSpace[rangeY - difference[:, None], rangeX]

    return alignedPinhole2_hv, difference, targetEdge


```
